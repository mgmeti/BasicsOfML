### Indexing & Slicing

1D Array

```python
arr = np.arange(10)

print(arr.shape)
print(arr.size)
print(arr.ndim)
print(arr.dtype)
print(arr.itemsize)
print("---------")
## Indexing
print(arr[2]) # access element at index 2
print(arr[0])
# print(arr[11]) # IndexError: index 11 is out of bounds for axis 0 with size 10
print(arr[-1]) # reverse access
print(arr[-2])
print("---------")

## Slicing
print(arr[2: 5]) # note: 5 is exclisive here
print(arr[:]) # everything
print(arr[2:]) # from 2 and to all
print(arr[:5]) # till 5
print(arr[::2]) # every 2nd element [0 2 4 6 8]
```

2D Array

```python
b = np.array([[10,20,30],
              [40,50,60],
              [70,80,90]])

print(b)
print(b.size, b.shape, b.ndim, b.dtype, b.itemsize, sep='\n')
print("---------")
print(b[0, 1])
print(b[2, 1])
print(b[:, 1]) # entire column
print(b[1, :]) # entire row
print(b[:2, :2])
```

Boolean Indexing

```python
a = np.array([10, 20, 30, 40, 50])
print(a[a>25])
print(a[a % 20 == 0])
```

Fancy Indexing

```python
a = np.arange(10, 90, 10)
print(a)
print(a[[0, 3, 5]]) # [10 40 60]
```

Reshaping Arrays

```python
a = np.arange(12)   # [0, ..., 11]
b = a.reshape(3, 4) # 3x4 matrix
print(a)
print(b)
print("----------")
print(b.flatten())      # convert back to 1D
print(b.ravel())        # similar but gives a view (linked to original)
```

Copy vs View

```python
a = np.arange(5) # [0,....., 4]
print("----------Direct reference---------")
b = a  # reference (no copy)
print(a, b, sep='\n')
b[0] = 100
print(a, b, sep='\n')
print("----------View of array ----------")
c = a.view() # shallow copy (linked data)
print(a,b, c, sep='\n')
c[0] = 200
print(a,b, c, sep='\n')
print("----------Copy of array ---------")
d = a.copy()  # deep copy (independent)
print(a,b, c, d, sep='\n')
d[0] = 1000
print(a, b, c, d, sep='\n')
```

## Module 3: Math & Statistics with NumPy

✅ Goal: Perform numerical operations efficiently.

    Element-wise operations (+ - * / **)

    Aggregate functions: sum, min, max, mean, std, var, median

    Axis-based operations (axis=0 vs axis=1)

    Broadcasting rules

### Element-wise Operations

NumPy automatically applies arithmetic operations to each element.

```python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

print(a + b)
print(b - a)
print(a - b)
print(a * b)
print(a / b)
print( b / a)
print(a ** b)
print(b ** a)
print( a ** 2)
```

### Aggregate Functions

```python
x = np.array([1, 2, 3, 4, 5])
print(x.sum())
print(x.min())
print(x.max())
print(x.mean())
print(x.std())
print(x.var())
print(np.median(x))
```

### Axis-based Operations

Think rows vs columns.

    axis=0 → operate column-wise

    axis=1 → operate row-wise

```python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print(m.sum(axis=0)) # column-wise
print(m.sum(axis=1)) # row-wise
```

### Broadcasting
NumPy lets you combine arrays of different shapes.

```python
a = np.array([1, 2, 3])
b = np.array([[10], [20], [30]])
print(a.shape, b.shape, sep='\n')
print(a, b, sep='\n')
print("Here, a (1D) automatically broadcasts to match b (2D).")
print(a + b)
```

### Normalization Example

```python
data = np.random.randint(1, 100, (3,3))
print("Original:\n", data)

mean = data.mean()
std = data.std()

print("Mean:", mean)
print("Standard Deviation:", std)
normalized = (data-mean)/std
print("Normalized:\n", normalized)
```

## Module 4: Linear Algebra with NumPy

✅ Goal: Work with matrices and vectors like in math.

    Dot product → np.dot / @ operator

    Matrix multiplication → np.matmul

    Transpose → .T

    Determinant, Inverse → np.linalg.det, np.linalg.inv

    Eigenvalues & eigenvectors → np.linalg.eig

### Dot Product & Matrix Multiplication

```python
a = np.array([1, 2])
b = np.array([3, 4])

print(np.dot(a, a))
print(np.dot(a, b))  # 1*3 + 2*4 = 11
print(a @ b) # matrix multiplication
print(np.dot(b, b))


print("-------- 2D --------")
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print(np.dot(A, B))   # same as @
print(A @ B) # matrix multiplication
```

### Transpose

```python
A = np.array([[1,2,3],[4,5,6]])
print(A)
print(A.shape)
print(A.T)
print(A.T.shape)
```

### Determinant & Inverse

```python
A = np.array([[1,2],[3,4]])

print(np.linalg.det(A)) # determinant
print(np.linalg.inv(A))   # inverse

print(np.linalg.det(A.T)) # determinant
print(np.linalg.inv(A.T))   # inverse
```

### Solving Linear Equations
Equation system:

    2x + y = 5  
    x + 3y = 10

Matrix form: Ax = b

```python
A = np.array([[2, 1], [1, 3]])
b = np.array([5, 10])

x = np.linalg.solve(A, b)
print(x)
```

### Eigenvalues & Eigenvectors

```python
A = np.array([[4, -2],
              [1,  1]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print(eigenvalues)
print(eigenvectors)
```

### Vectorization (avoid Python loops)
Use array operations / ufuncs instead of Python for loops — much faster.

```python
import numpy as np

# BAD (slow)
x = np.arange(1_000_000)
out_loop = np.empty_like(x)

for i in range(len(x)):
    out_loop[i] = x[i]**2 + 2*x[i] + 1

print(out_loop)

# GOOD (vectorized, fast)
out_vector = x**2 + 2*x + 1
print(out_vector)
```

### Universal functions (ufuncs) & np.where
Built-in vectorized functions (sin, exp, sqrt, log, etc.). Use np.where for conditional selection.

```python
y = np.sin(x) # elementwise

z = np.where( x % 2 == 0, x // 2, -1) # conditional per-element
```

### einsum — concise, fast linear algebra and reductions
Powerful for contractions, sums, dot products, transposes without temporaries.

```python
A = np.random.rand(3,4)
B = np.random.rand(4,5)
C = np.einsum('ij,jk->ik', A, B)  # A @ B
# sum over rows:
s = np.einsum('ij->i', A)         # row sums

print(C, s, sep='\n')
```

### Advanced broadcasting tricks
Add a column vector to every row, or compute pairwise distances without loops.

```python
M = np.random.rand(100, 3)
# print(M)
col_mean = M.mean(axis=0)         # shape (3,)
print(col_mean)
M_centered = M - col_mean        # broadcasted
print(M_centered)
```

```python
# pairwise squared distances (N x N)
X = np.random.rand(100, 2)
d2 = np.sum((X[:, None, :] - X[None, :, :])**2, axis=-1)
print(d2)
```

### Structured / record arrays
Store heterogeneous columns (like table rows) in an array.

```python
dt = np.dtype([('name', 'U10'), ('age', 'i4'), ('score', 'f4')])
arr = np.array([('Amy', 30, 9.1), ('Bob', 25, 7.3)], dtype=dt)
print(arr['score'])   # access column
```

```python
rng = np.random.default_rng(42)
data = rng.normal(loc=0, scale=1, size=(1000,))
print(data)
```

### I/O: np.save, np.load, savetxt, loadtxt
Save and load efficiently (binary .npy is fastest).

```python
np.save('arr.npy', arr)
arr2 = np.load('arr.npy', mmap_mode=None)  # or mmap_mode='r' to memmap
print(arr)
print(arr2)
```

```python
# Linear regression via normal equations (small, but instructive):
# X: (N, p) design, y: (N, )
X = np.random.rand(100, 3)
y = np.random.rand(100)
X = np.hstack([np.ones((100, 1)), X])   # add intercept
beta = np.linalg.solve(X.T @ X, X.T @ y) # normal equations
print(beta)
```

### NumPy + Pandas (Data Wrangling)
Pandas builds on top of NumPy. Its Series and DataFrame objects internally use NumPy arrays.

```python
import numpy as np
import pandas as pd

# Array
data = np.array([[10, 20, 30], [40, 50 ,60], [70, 80, 90]])

# Create Panda dataframe from Array
df = pd.DataFrame(data, columns=['A', 'B', 'C'])

print(df)

print("Column 'B' mean", df['B'].mean())  # Uses NumPy under the hood
print("Row sum: ", df.sum(axis=1).to_numpy)
```

### NumPy + Matplotlib (Plotting)
Matplotlib often takes NumPy arrays as input for plotting.

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)   # 100 values between 0 and 2π
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.title("NumPy + Matplotlib Example")
plt.xlabel("x values")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
```

### NumPy + SciPy (Scientific Computing)
SciPy builds on NumPy for advanced math, stats, and optimization.

```python
from scipy import linalg

A = np.array([[2, 3], [1, 4]])
det_A = linalg.det(A)
eigvals, eigvecs = linalg.eig(A)

print("Determinant:", det_A)
print("Eigenvalues:", eigvals)
```

### NumPy in Machine Learning
Most ML libraries (Scikit-learn, TensorFlow, PyTorch) use NumPy arrays as their backbone.

```python
from sklearn.linear_model import LinearRegression

# Example dataset
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

print("Prediction for 6:", model.predict([[6]]))  # Should be ~12
```
